## Preparing data for the missing-value tasks
The BEPS dataset is **complete (no missing values)**, so to practise tasks 8-18 we create a copy `df_m` in which ~5% of the values in a few columns are randomly set to `NaN` (fixed seed -> reproducible).
**Delete this cell if your own dataset already has missing values** and use `df_m = df.copy()`.

```python
rng = np.random.default_rng(42)
df_m = df.copy()

num_cols = df_m.select_dtypes(include="number").columns.tolist()
df_m[num_cols] = df_m[num_cols].astype(float)         # so NaN can be stored in numeric columns

for col in ["age", "Blair", "Hague", "Europe", "vote", "gender"]:
    idx = rng.choice(df_m.index, size=int(0.05 * len(df_m)), replace=False)
    df_m.loc[idx, col] = np.nan

print("Missing values introduced:")
print(df_m.isna().sum())
```

## 12. Remove all rows containing missing values

```python
df_dropna = df_m.dropna()
print("Before:", df_m.shape, "| After:", df_dropna.shape)
df_dropna.head(10)
```

## 13. Replace missing values of a numerical column with the **mean**

```python
df_work = df_m.copy()                       # working copy used for the remaining steps
df_work["age"] = df_work["age"].fillna(df_work["age"].mean())
print("Missing in age:", df_work["age"].isna().sum())
```

## 14. Replace missing values of another numerical column with the **median**

```python
df_work["Blair"] = df_work["Blair"].fillna(df_work["Blair"].median())
print("Missing in Blair:", df_work["Blair"].isna().sum())
```

## 15. Replace missing values of a categorical column with the **mode**

```python
mode_vote = df_work["vote"].mode()[0]
df_work["vote"] = df_work["vote"].fillna(mode_vote)
print("Mode of vote:", mode_vote, "| Missing in vote:", df_work["vote"].isna().sum())
```

## 16. Forward fill (last observation carried forward - R `zoo::na.locf()`)

```python
df_work["Hague"] = df_work["Hague"].ffill()
print("Missing in Hague after ffill:", df_work["Hague"].isna().sum())
```

## 17. Backward fill (next observation carried backward)

```python
df_work["Europe"] = df_work["Europe"].bfill()
print("Missing in Europe after bfill:", df_work["Europe"].isna().sum())
```

## 18. Replace remaining missing values with a constant (0 / "Unknown")

```python
print("Remaining missing before:\n", df_work.isna().sum()[df_work.isna().sum() > 0])

num_c = df_work.select_dtypes(include="number").columns
cat_c = df_work.columns.difference(num_c)
df_work[num_c] = df_work[num_c].fillna(0)
df_work[cat_c] = df_work[cat_c].fillna("Unknown")

print("\nRemaining missing after:", df_work.isna().sum().sum())
```

## 19. Remove duplicate records

```python
print("Duplicate rows found:", df_work.duplicated().sum())

# BEPS may have no exact duplicates -> add a few copies purely to demonstrate the removal
if df_work.duplicated().sum() == 0:
    df_work = pd.concat([df_work, df_work.sample(10, random_state=1)], ignore_index=True)
    print("Added 10 duplicate rows for demonstration; duplicates now:", df_work.duplicated().sum())

df_work = df_work.drop_duplicates().reset_index(drop=True)
print("Duplicates after removal:", df_work.duplicated().sum(), "| Shape:", df_work.shape)
```

## 20. Convert appropriate variables into factors (categorical)

```python
for col in ["vote", "gender"]:
    df_work[col] = df_work[col].astype("category")

# ordinal 1-5 rating columns can also be treated as ordered factors
for col in ["economic_cond_national", "economic_cond_household"]:
    df_work[col] = pd.Categorical(df_work[col].astype(int), ordered=True)

print(df_work.dtypes)
print("\nLevels of vote:", list(df_work["vote"].cat.categories))
```

## 21. Rename one or more columns

```python
df_work = df_work.rename(columns={"economic_cond_national": "econ_national",
                                  "economic_cond_household": "econ_household"})
print(df_work.columns.tolist())
```

## 22. Create a new derived column

```python
df_work["econ_avg"] = (df_work["econ_national"].astype(int) + df_work["econ_household"].astype(int)) / 2
df_work["age_group"] = pd.cut(df_work["age"], bins=[0, 30, 45, 60, 120],
                              labels=["<=30", "31-45", "46-60", "60+"])
df_work[["age", "age_group", "econ_national", "econ_household", "econ_avg"]].head(10)
```

## 24-28. Min, Max, Range, Variance, Standard Deviation, IQR (variable: `age`)

```python
x = df_work["age"]
print("24. Minimum :", x.min())
print("    Maximum :", x.max())
print("25. Range   :", x.max() - x.min())
print("26. Variance:", round(x.var(), 4))           # sample variance (ddof=1, same as R)
print("27. Std Dev :", round(x.std(), 4))
q1, q3 = x.quantile(0.25), x.quantile(0.75)
print("28. IQR     :", q3 - q1, f"(Q1={q1}, Q3={q3})")
```
