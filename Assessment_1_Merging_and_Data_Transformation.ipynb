## Part A - Dataframe Operations

```python
import numpy as np
import pandas as pd
import re
from IPython.display import display
```

### Create dataframe1, dataframe2, dataframe3 (10 records each)


```python
dataframe1 = pd.DataFrame({
    "ProductID":  range(101, 111),
    "NameofProd": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer",
                   "Scanner", "Webcam", "Headset", "Router", "Tablet"],
    "Price":      [55000, 500, 1200, 12000, 8500, 6500, 2500, 3000, 2200, 22000]
})

dataframe2 = pd.DataFrame({
    "ProductID":        range(106, 116),
    "Company_Location": ["Chennai", "Bengaluru", "Hyderabad", "Mumbai", "Delhi",
                         "Pune", "Kolkata", "Coimbatore", "Ahmedabad", "Jaipur"]
})

dataframe3 = pd.DataFrame({
    "ProductID":  range(111, 121),
    "NameofProd": ["Speaker", "Pendrive", "Hard Disk", "SSD", "Projector",
                   "Smartwatch", "Charger", "Power Bank", "Earbuds", "Camera"],
    "Price":      [3500, 600, 4500, 5200, 30000, 7000, 900, 1800, 2000, 38000]
})

for name, d in [("dataframe1", dataframe1), ("dataframe2", dataframe2), ("dataframe3", dataframe3)]:
    print(name, d.shape)
    display(d)
```

### 1. Append the records of dataframe1 and dataframe3

```python
appended = pd.concat([dataframe1, dataframe3], ignore_index=True)
print(appended.shape)
display(appended)
```

### 2. Concatenate the data frames wherever applicable


```python
# Row-wise concatenation (same columns)
row_concat = pd.concat([dataframe1, dataframe3], axis=0, ignore_index=True)
print("Row-wise concat of dataframe1 + dataframe3:", row_concat.shape)

# Column-wise concatenation (same number of rows, rows are placed side by side by position)
col_concat = pd.concat([dataframe1, dataframe2], axis=1)
print("\nColumn-wise concat of dataframe1 + dataframe2:", col_concat.shape)
display(col_concat)

# Note: dataframe2 + dataframe3 row-wise is NOT meaningful (different columns -> NaN filled)
display(pd.concat([dataframe2, dataframe3], axis=0, ignore_index=True).head())
```

### 3. Merge dataframe1 and dataframe2 using ProductID

```python
merged = pd.merge(dataframe1, dataframe2, on="ProductID")
display(merged)
```

### 4. Join operations between dataframe1 and dataframe2

```python
print("INNER JOIN (only matching ProductIDs)")
display(pd.merge(dataframe1, dataframe2, on="ProductID", how="inner"))

print("LEFT JOIN (all rows of dataframe1)")
display(pd.merge(dataframe1, dataframe2, on="ProductID", how="left"))

print("RIGHT JOIN (all rows of dataframe2)")
display(pd.merge(dataframe1, dataframe2, on="ProductID", how="right"))

print("FULL OUTER JOIN (all rows of both)")
display(pd.merge(dataframe1, dataframe2, on="ProductID", how="outer"))
```

## Part B - Data Transformations


```python
FIRST_NAME = "rahul"        # <-- change to YOUR first name
REG_NO     = "23BCE1234"    # <-- change to YOUR registration number

last4 = int("".join(re.findall(r"\d", REG_NO))[-4:])      # last four digits of the Reg. No.
values = [last4, 12, 45, 7, 33, 58, 21, 90, 64, 5]         # first value = last 4 digits + nine integers

globals()[FIRST_NAME] = np.array(values, dtype=float)      # object named after the first name
Fname = globals()[FIRST_NAME]                               # handy alias used below

print(f"Vector '{FIRST_NAME}':", Fname)
```

### 1. Normalization to [-3, +3], [-1, +1] and [0, 1] (Min-Max)
Formula: `x' = a + (x - min)(b - a) / (max - min)`

```python
def rescale(x, a, b):
    return a + (x - x.min()) * (b - a) / (x.max() - x.min())

norm_3_3   = rescale(Fname, -3, 3)
norm_1_1   = rescale(Fname, -1, 1)
minmax_0_1 = rescale(Fname,  0, 1)

print("[-3, +3]:", np.round(norm_3_3, 4))
print("[-1, +1]:", np.round(norm_1_1, 4))
print("[ 0,  1]:", np.round(minmax_0_1, 4))
```

### 2. Z-score: `z = (x - mean) / sd`  (sample SD, ddof=1, same as R's `scale()`)

```python
zscore = (Fname - Fname.mean()) / Fname.std(ddof=1)
print("Z-score:", np.round(zscore, 4))
print("Mean ~", round(zscore.mean(), 6), "| SD =", round(zscore.std(ddof=1), 6))
```

### 3. Decimal Scaling: `x' = x / 10^j`, where j is the smallest integer such that max(|x'|) < 1

```python
j = 0
while np.max(np.abs(Fname)) / (10 ** j) >= 1:
    j += 1
decimal_scaled = Fname / (10 ** j)
print("j =", j, "-> divide by", 10 ** j)
print("Decimal scaling:", np.round(decimal_scaled, 4))
```

### Summary table of all transformations

```python
summary = pd.DataFrame({
    "Original": Fname,
    "Norm[-3,3]": norm_3_3,
    "Norm[-1,1]": norm_1_1,
    "MinMax[0,1]": minmax_0_1,
    "Z-score": zscore,
    "DecimalScaling": decimal_scaled
}).round(4)
display(summary)
```
